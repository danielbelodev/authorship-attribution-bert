```python
import pandas as pd
from dateutil import parser

# Custom function to parse each date string
def parse_dates(date_str):
    try:
        # Use dateutil.parser to handle timezone and convert to UTC
        return parser.parse(date_str)
    except Exception:
        return pd.NaT

# Apply the function to the 'lastmod' column
df['lastmod'] = df['lastmod'].apply(parse_dates)

# Print the result to check if the conversion works
print(df['lastmod'])
```
